```python
import pandas as pd
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.tail()
```

```python
df.shape
```

```python
df.describe()
```

```python
df.groupby('time')['tip'].mean()
```

```python
df.groupby('time')['total_bill'].mean()
```

```python
df['tip_pct'] = df['tip'] / df['total_bill'] * 100
```

```python
df.groupby('time')['tip_pct'].mean()
```

```python
df.groupby('smoker')['tip_pct'].mean()
```

```python
df.groupby('sex')['tip_pct'].mean()
```

This dataset provides an overview of 244 rows and 7 column of restaurant vists. it looked at the total bill amount, the tip amount and othe factors that might have influenced tipping(time - whether the visit was lunch or dinner, smoking - whether the person smokes or not, the size of the of the party, and gender). After the analysis dinner tip was $3.10 while the lunch tip was $2.73 but upon applying percentage to it was seen that the lunch tip is slightly higher than the dinner tip.Gender showed a clear difference; female customers tipped about 16.6% on averag eversus 15.8% for male customers.  Smokers also tend to tip slightly higher than non smokers though this gap is small enough that it may not be a meaningful difference given the sample size
